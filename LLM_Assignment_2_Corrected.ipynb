```python
import torch
import torch.nn as nn
import torch.optim as optim
import string

"""
Why these imports?
- torch, torch.nn, torch.optim : Core PyTorch for building and training model
- string                       : To get all letters, digits, punctuation for vocabulary
"""
```

```python
# ==================== Hyperparameters ====================

embedding_dim = 32
hidden_dim = 64
num_layers = 2
learning_rate = 0.005
epochs = 10
seq_length = 50          # Length of input sequence for training

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
# ==================== Load Your Training Text Here ====================

# Paste your ~1000 words text here
text = """
**A Concise Overview of Indian History**

India possesses one of the world’s oldest continuous civilizations, with a recorded history spanning more than five thousand years. Located at the crossroads of ancient trade routes connecting Central Asia, Southeast Asia, and the Middle East, the Indian subcontinent has been a crucible of cultural exchange, religious innovation, imperial ambition, and intellectual achievement. Its story encompasses the rise and fall of great empires, the birth of major world religions, remarkable advances in science and the arts, and a long struggle for independence that shaped the world’s largest democracy. This overview presents the major periods and themes of Indian history in clear chronological order.

**Ancient India (c. 3300 BCE – 550 CE)**

The earliest urban civilization on the subcontinent emerged in the Indus Valley around 3300 BCE. Major sites such as Harappa and Mohenjo-daro, now in Pakistan, reveal remarkably sophisticated urban planning: grid-patterned streets, covered drainage systems, standardized bricks, public baths, and granaries. The civilization maintained extensive trade links with Mesopotamia, exporting cotton textiles, carnelian beads, and pottery. Its script remains undeciphered, and scholars continue to debate the causes of its gradual decline after 1900 BCE, which may have involved climate shifts, river course changes, and population movements.

Around 1500 BCE, Indo-Aryan communities migrated into northern India, bringing with them the Sanskrit language and composing the four Vedas. The Vedic period laid the foundations of early Hinduism, including concepts of dharma, karma, and the varna social order. Philosophical speculation deepened in the Upanishads. By the sixth century BCE, sixteen large territorial states known as Mahajanapadas had emerged in the Gangetic plain. It was in this environment that two influential heterodox traditions arose: Buddhism, founded by Siddhartha Gautama (the Buddha), who taught the Four Noble Truths and the path to end suffering; and Jainism, associated with Mahavira, which emphasized strict non-violence and ascetic discipline. Both movements challenged ritualistic excess and promoted ethical living accessible to all social groups.

The first major empire to unify much of the subcontinent was the Mauryan Empire (322–185 BCE). Chandragupta Maurya, guided by the brilliant strategist Chanakya (Kautilya), overthrew the Nanda dynasty and established a centralized administration whose principles are recorded in the Arthashastra, an early treatise on statecraft, economics, and espionage. Chandragupta’s grandson Ashoka expanded the empire dramatically but was profoundly affected by the violence of the Kalinga War around 261 BCE. He embraced Buddhism and inscribed edicts on rocks and pillars across his realm promoting dhamma—tolerance, welfare measures, and non-violence. Ashoka’s missionaries carried Buddhism to Sri Lanka, Central Asia, and Southeast Asia.

After the Mauryas, the subcontinent saw regional powers rise and fall. The Kushana Empire in the northwest, especially under Kanishka (c. 127–150 CE), facilitated Silk Road trade and patronized Buddhism and Greek-influenced art known as Gandhara. In the Deccan, the Satavahanas promoted trade and issued some of the earliest inscribed coins. The Gupta Empire (c. 320–550 CE) is often described as a golden age. Mathematician Aryabhata advanced astronomy and proposed that the Earth rotates on its axis. The decimal place-value system and the concept of zero as a numeral matured during this period. Kalidasa produced enduring Sanskrit literature, including the play Abhijnanashakuntalam. Cave temples at Ajanta and Ellora display exquisite Buddhist, Hindu, and Jain art. Medicine advanced through the compilations attributed to Charaka and Sushruta. In southern India, the Chera, Chola, and Pandya kingdoms flourished; Sangam Tamil literature vividly describes ancient society, long-distance trade with Rome, and maritime activity.

**Medieval India (c. 550 – 1750 CE)**

The early medieval centuries witnessed the proliferation of regional kingdoms and the flowering of temple architecture and devotional movements. The Pallavas of Kanchipuram pioneered monolithic rock-cut temples at Mahabalipuram. The Chalukyas and Rashtrakutas left magnificent monuments at Badami and Ellora. The Chola Empire reached its height under Rajaraja I and Rajendra I (10th–11th centuries), building the enormous Brihadeeswarar Temple at Thanjavur, developing sophisticated irrigation and village administration, and launching naval expeditions that carried Indian cultural influence to Southeast Asia.

From the late 10th century, Turkish and Afghan military campaigns began to reshape northern India. Mahmud of Ghazni conducted repeated raids in the early 11th century. In 1206, Qutb ud-Din Aibak founded the Delhi Sultanate, which endured until 1526. Successive dynasties—the Mamluks, Khaljis, Tughlaqs, Sayyids, and Lodis—extended Muslim rule across much of the north and into the Deccan. Alauddin Khalji introduced market controls and expanded southward. Muhammad bin Tughlaq attempted ambitious but often impractical administrative reforms. The Sultanate period introduced Persian as the language of administration and high culture, new revenue systems, and a distinctive Indo-Islamic architectural style visible in the Qutub Minar complex and Alai Darwaza.

Alongside political change, the Bhakti movement spread across the subcontinent from roughly the 7th century onward. Poet-saints such as Kabir, Ravidas, Mirabai, and Tulsidas emphasized personal devotion, criticized caste hierarchies, and composed verses in regional languages. In Punjab, Guru Nanak founded Sikhism in the late 15th century, stressing equality, honest work, and remembrance of the divine. Sufi saints, including Nizamuddin Auliya and Khwaja Moinuddin Chishti, promoted spiritual love and attracted followers from many communities, contributing to a composite cultural tradition in music, poetry, and daily life.

In the south, the Vijayanagara Empire (1336–1646) emerged as a major center of Hindu political and cultural resistance. Its capital at Hampi became a wealthy cosmopolitan city until its destruction in 1565. Meanwhile, the Mughal Empire was founded in 1526 when Babur defeated Ibrahim Lodi at the First Battle of Panipat. Babur’s grandson Akbar (r. 1556–1605) proved an exceptionally capable ruler. He abolished the jizya tax on non-Muslims, pursued a policy of religious tolerance (sulh-i-kul), created the efficient Mansabdari administrative system, and patronized scholars and artists of all faiths. His court at Fatehpur Sikri and the later monuments built by Shah Jahan, including the Taj Mahal, represent the peak of Indo-Islamic architecture and court culture. Aurangzeb (r. 1658–1707) expanded the empire to its greatest territorial extent but faced prolonged wars in the Deccan and growing resistance from Marathas under Shivaji, Sikhs, and others. By the early 18th century, Mughal central authority had weakened, creating opportunities for regional powers and European trading companies.

**Colonial Rule and the Freedom Struggle (1757 – 1947)**

European interest in India intensified after Vasco da Gama reached Calicut in 1498. The Portuguese established a presence in Goa, the Dutch in parts of the east coast, and the French at Pondicherry. The British East India Company, chartered in 1600, gradually became the dominant European power. Its victory at the Battle of Plassey in 1757, achieved with local allies against Nawab Siraj-ud-Daulah of Bengal, marked the beginning of territorial expansion. Through a combination of military campaigns, subsidiary alliances, and the Doctrine of Lapse, the Company brought most of the subcontinent under its control by the mid-19th century.

The Revolt of 1857, also known as the First War of Independence or Sepoy Mutiny, erupted over military grievances but reflected deeper discontent with economic exploitation, cultural interference, and rapid annexation. Leaders included Rani Lakshmibai of Jhansi, Nana Saheb, and Bahadur Shah Zafar. Although the uprising was suppressed, it ended Company rule. The British Crown assumed direct governance in 1858. Queen Victoria’s Proclamation promised religious non-interference and equal treatment under law.

Modern Indian nationalism emerged in the late 19th century. The Indian National Congress was founded in 1885. Early leaders such as Gopal Krishna Gokhale favored constitutional agitation, while Bal Gangadhar Tilak demanded swaraj (self-rule) and promoted the Swadeshi movement after the partition of Bengal in 1905. Mohandas Karamchand Gandhi returned from South Africa in 1915 and introduced satyagraha—mass non-violent civil disobedience—into the national movement. Landmark campaigns included the Non-Cooperation Movement (1920–22), the Civil Disobedience Movement highlighted by the 1930 Dandi Salt March, and the Quit India Movement of 1942. Parallel streams included revolutionary nationalists such as Bhagat Singh and Subhas Chandra Bose’s Indian National Army.

The All-India Muslim League, under Muhammad Ali Jinnah, increasingly advocated the “two-nation theory” and demanded a separate homeland for Muslims. Despite last-minute negotiations, Partition accompanied independence. On 15 August 1947, India became a sovereign nation, but the division of Punjab and Bengal triggered horrific communal violence and one of the largest migrations in history, with millions displaced and an estimated one to two million killed.

**Independent India (1947 – Present)**

The Constitution of India, drafted under the leadership of Dr. B.R. Ambedkar and adopted in 1950, established a sovereign, secular, democratic republic with fundamental rights, universal adult franchise, and special provisions for historically disadvantaged groups. Sardar Vallabhbhai Patel oversaw the integration of more than 560 princely states into the Union. Jawaharlal Nehru, India’s first Prime Minister, pursued planned economic development through Five-Year Plans, promoted the Non-Aligned Movement in foreign policy, and invested in scientific and educational institutions.

India fought wars with Pakistan in 1947–48, 1965, and 1971 (the last resulting in the creation of Bangladesh) and with China in 1962. The Green Revolution of the 1960s and 1970s dramatically increased agricultural production. In 1991, facing a severe balance-of-payments crisis, the government initiated economic liberalization under Prime Minister P.V. Narasimha Rao and Finance Minister Manmohan Singh. The reforms unleashed rapid growth in information technology, services, and manufacturing, transforming India into one of the world’s fastest-growing major economies.

Today India is the world’s most populous country and its largest democracy, conducting elections with over 900 million eligible voters. It possesses nuclear weapons, maintains a professional armed force, and has achieved notable successes in space exploration, pharmaceuticals, and digital technology. At the same time, it continues to grapple with challenges of poverty reduction, inequality, communal harmony, environmental sustainability, and the management of extraordinary linguistic, religious, and regional diversity.

**Conclusion**

Indian history reveals a remarkable pattern of continuity amid change. Successive waves of migration, invasion, and trade brought new ideas, technologies, and peoples, yet Indian civilization repeatedly absorbed, adapted, and synthesized these influences while preserving distinctive philosophical, artistic, and social traditions. From the urban sophistication of the Indus Valley to the democratic vitality of the present day, India’s journey offers enduring lessons about resilience, pluralism, and the human capacity for both creativity and conflict. Understanding this long and complex past remains essential for anyone seeking to comprehend contemporary India and its evolving role in the world. 

"""

# Character-level vocabulary
chars = sorted(list(set(text)))
vocab_size = len(chars)

# Do the tokenization
char_to_id = {ch: idx for idx, ch in enumerate(chars)}
id_to_char = {idx: ch for idx, ch in enumerate(chars)}

print(char_to_id)
print()
print(id_to_char)
print(f"Vocabulary size (unique characters): {vocab_size}")
```

```python
# ==================== Create Training Dataset ====================

def create_dataset(text, seq_length):
    input_sequences = []
    target_sequences = []
    
    for i in range(len(text) - seq_length):
        input_seq = text[i:i + seq_length]
        target_seq = text[i + 1:i + seq_length + 1]
        
        input_sequences.append([char_to_id[ch] for ch in input_seq])
        target_sequences.append([char_to_id[ch] for ch in target_seq])
    
    return torch.tensor(input_sequences, dtype=torch.long), torch.tensor(target_sequences, dtype=torch.long)

inputs, targets = create_dataset(text, seq_length)

print(inputs[:2])
print(f"Total training examples: {len(inputs)}")
```

```python
# ==================== Character-Based Nano LLM ====================

class CharNanoLLM(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, num_layers):
        super().__init__()
        
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        
        self.lstm = nn.LSTM(
            embedding_dim, 
            hidden_dim, 
            num_layers=num_layers, 
            batch_first=True
        )
        
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x, hidden=None):
        embedded = self.embedding(x)
        lstm_out, hidden = self.lstm(embedded, hidden)
        logits = self.fc(lstm_out)
        return logits, hidden

model = CharNanoLLM(vocab_size, embedding_dim, hidden_dim, num_layers).to(device)
print("Character-based Nano LLM created.")
```

```python
# ==================== Training Loop (Improved) ====================

from tqdm import tqdm

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

print("\n" + "="*60)
print("TRAINING STARTED")
print("="*60 + "\n")

for epoch in range(1, epochs + 1):
    
    model.train()
    total_loss = 0.0
    num_batches = 0
    
    # Progress bar for each epoch
    progress_bar = tqdm(
        range(0, len(inputs), 32), 
        desc=f"Epoch {epoch}/{epochs}", 
        leave=False
    )
    
    for i in progress_bar:
        batch_inputs = inputs[i:i+32].to(device)
        batch_targets = targets[i:i+32].to(device)
        
        optimizer.zero_grad()
        
        outputs, _ = model(batch_inputs)
        loss = criterion(outputs.view(-1, vocab_size), batch_targets.view(-1))
        
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        num_batches += 1
        
        # Update progress bar with current loss
        progress_bar.set_postfix({"Loss": f"{loss.item():.4f}"})
    
    # Calculate average loss for the epoch
    avg_loss = total_loss / num_batches
    
    # Print epoch summary
    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{epochs} | Average Loss: {avg_loss:.4f}")

print("\n" + "="*60)
print("TRAINING COMPLETED")
print("="*60 + "\n")
```

```python
# ==================== Text Generation ====================

def generate_text(model, start_text, length=200, temperature=0.8):
    model.eval()
    generated = start_text
    
    # Convert starting text to tensor
    input_seq = torch.tensor([char_to_id.get(ch, 0) for ch in start_text]).unsqueeze(0).to(device)
    hidden = None
    
    with torch.no_grad():
        for _ in range(length):
            output, hidden = model(input_seq, hidden)
            output = output[:, -1, :] / temperature
            probs = torch.softmax(output, dim=-1)
            next_char_idx = torch.multinomial(probs, num_samples=1).item()
            
            next_char = id_to_char.get(next_char_idx, " ")
            generated += next_char
            
            # Update input for next prediction
            input_seq = torch.tensor([[next_char_idx]]).to(device)
    
    return generated


# ==================== User Input + Generation ====================

print("\n=== Character-Based Text Generator ===")
prompt = input("Enter starting text (e.g., 'Ancient India'): ")

generated = generate_text(model, prompt, length=150, temperature=0.7)
print("\nGenerated Text:\n")
print(generated)
```
